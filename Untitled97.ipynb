Apply ARIMA to Check Evaluation

```python
import numpy as np
from statsmodels.tsa.arima.model import ARIMA
from sklearn.metrics import mean_squared_error

# --------------------------------------------------
# 1. Target variable
# --------------------------------------------------

y = df['Unemployment_Rate(%)'].dropna()


# --------------------------------------------------
# 2. 80/20 Train-Test Split
# --------------------------------------------------

split_point = int(len(y) * 0.80)

train = y.iloc[:split_point]
test = y.iloc[split_point:]

print("Train observations:", len(train))
print("Test observations:", len(test))


# --------------------------------------------------
# 3. ARIMA(0,1,1)
# --------------------------------------------------

model = ARIMA(
    train,
    order=(0, 1, 1)
)

result = model.fit()

print(result.summary())


# --------------------------------------------------
# 4. Train Prediction
# --------------------------------------------------

train_pred = result.predict(
    start=train.index[0],
    end=train.index[-1]
)

train_pred = train_pred.reindex(train.index)

train_rmse = np.sqrt(
    mean_squared_error(train, train_pred)
)


# --------------------------------------------------
# 5. Test Forecast
# --------------------------------------------------

test_pred = result.forecast(
    steps=len(test)
)

test_pred = np.asarray(test_pred)

test_rmse = np.sqrt(
    mean_squared_error(test, test_pred)
)


# --------------------------------------------------
# 6. Results
# --------------------------------------------------

print("\n===== ARIMA(0,1,1) | 80/20 Split =====")

print(f"Train RMSE : {train_rmse:.4f}")
print(f"Test RMSE  : {test_rmse:.4f}")

print(f"AIC        : {result.aic:.4f}")
print(f"BIC        : {result.bic:.4f}")
```

```python
import matplotlib.pyplot as plt
from statsmodels.graphics.tsaplots import plot_acf
from statsmodels.stats.diagnostic import acorr_ljungbox

# --------------------------------------------------
# 1. Residuals
# --------------------------------------------------

residuals = result.resid.dropna()


# --------------------------------------------------
# 2. Residual Plot
# --------------------------------------------------

plt.figure(figsize=(10, 4))

plt.plot(residuals)

plt.axhline(
    y=0,
    linestyle='--'
)

plt.title('Residuals - ARIMA(0,1,1)')
plt.xlabel('Time')
plt.ylabel('Residuals')

plt.show()


# --------------------------------------------------
# 3. Residual ACF
# --------------------------------------------------

plt.figure(figsize=(10, 4))

plot_acf(
    residuals,
    lags=20,
    zero=False
)

plt.title('ACF of ARIMA(0,1,1) Residuals')

plt.show()


# --------------------------------------------------
# 4. Ljung-Box Test
# --------------------------------------------------

ljung_box = acorr_ljungbox(
    residuals,
    lags=[10, 15, 20],
    return_df=True
)

print("\nLjung-Box Test:")
print(ljung_box)
```

```python
import pandas as pd
from statsmodels.tsa.arima.model import ARIMA

# Target variable
y = df['Unemployment_Rate(%)'].dropna()

# Final ARIMA(0,1,1) model
final_model = ARIMA(
    y,
    order=(0,1,1)
)

final_result = final_model.fit()

# Forecast next 5 years
forecast_result = final_result.get_forecast(steps=5)

forecast_mean = forecast_result.predicted_mean
forecast_ci = forecast_result.conf_int()

# Create future years
future_years = pd.date_range(
    start='2026',
    periods=5,
    freq='YS'
)

# Forecast table
forecast_df = pd.DataFrame({
    'Year': future_years.year,
    'Forecast_Unemployment_Rate': forecast_mean.values,
    'Lower_95%': forecast_ci.iloc[:, 0].values,
    'Upper_95%': forecast_ci.iloc[:, 1].values
})

print(forecast_df)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(12, 5))

# Actual data
plt.plot(
    df['Year'],
    df['Unemployment_Rate(%)'],
    label='Actual'
)

# Forecast
plt.plot(
    forecast_df['Year'],
    forecast_df['Forecast_Unemployment_Rate'],
    marker='o',
    label='Forecast'
)

# 95% confidence interval
plt.fill_between(
    forecast_df['Year'],
    forecast_df['Lower_95%'],
    forecast_df['Upper_95%'],
    alpha=0.2,
    label='95% Confidence Interval'
)

plt.xlabel('Year')
plt.ylabel('Unemployment Rate (%)')
plt.title('Unemployment Rate Forecast using ARIMA(0,1,1)')
plt.legend()
plt.grid(False)

plt.show()
```
